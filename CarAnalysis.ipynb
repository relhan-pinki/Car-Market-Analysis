```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Car Dekho Data.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

1. Average Selling Price by Fuel Type

```python
plt.figure(figsize=(8, 5))

avg_fuel_price = (
    df.groupby("Fuel_Type")["Selling_Price"]
      .mean()
      .sort_values(ascending=False)
)

sns.barplot(
    x=avg_fuel_price.index,
    y=avg_fuel_price.values
)

plt.title("Average Selling Price by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Average Selling Price (Lakhs)")
plt.show()
```

2. Average Selling Price by Transmission

```python
plt.figure(figsize=(7, 5))

avg_transmission_price = (
    df.groupby("Transmission")["Selling_Price"]
      .mean()
      .sort_values(ascending=False)
)

sns.barplot(
    x=avg_transmission_price.index,
    y=avg_transmission_price.values
)

plt.title("Average Selling Price by Transmission")
plt.xlabel("Transmission")
plt.ylabel("Average Selling Price (Lakhs)")
plt.show()
```

3. Top 5 Car Models by Average Selling Price

```python
top_cars = (df.groupby("Car_Name")["Selling_Price"].mean().sort_values(ascending=False).head(5).sort_values(ascending=False))

plt.figure(figsize=(9, 6))

sns.barplot(
    x=top_cars.values,
    y=top_cars.index
)

plt.title("Top 5 Car Models by Average Selling Price")
plt.xlabel("Average Selling Price (Lakhs)")
plt.ylabel("Car Model")
plt.show()
```

4. Average Selling Price by Seller Type

```python
plt.figure(figsize=(7, 5))

seller_price = (df.groupby("Seller_Type")["Selling_Price"].mean().sort_values(ascending=False))

sns.barplot(
    x=seller_price.index,
    y=seller_price.values
)

plt.title("Average Selling Price by Seller Type")
plt.xlabel("Seller Type")
plt.ylabel("Average Selling Price (Lakhs)")
plt.show()
```

5. Present Price vs Selling Price

```python
plt.figure(figsize=(8,6))

sns.regplot(
    data=df,
    x="Present_Price",
    y="Selling_Price",
    scatter_kws={"alpha": 0.6},
    line_kws={"linewidth": 2},
    ci=None
)

plt.title("Present Price vs Selling Price")
plt.xlabel("Present Price (Lakhs)")
plt.ylabel("Selling Price (Lakhs)")
plt.tight_layout()
plt.show()
```

6. Average Selling price by manufacturing year

```python
year_price = (df.groupby("Year")["Selling_Price"].mean().reset_index())

plt.figure(figsize=(10, 5))

sns.lineplot(
    data=year_price,
    x="Year",
    y="Selling_Price",
    marker="o"
)

plt.title("Manufacturing Year vs Average Selling Price")
plt.xlabel("Manufacturing Year")
plt.ylabel("Average Selling Price (Lakhs)")
plt.show()
```

7. Kms Driven vs Selling Price

```python
plt.figure(figsize=(8,6))

sns.regplot(
    data=df,
    x="Kms_Driven",
    y="Selling_Price",
    scatter_kws={"alpha": 0.6},
    line_kws={"linewidth": 2},
    ci=None
)

plt.title("Kms Driven vs Selling Price")
plt.xlabel("Kilometers Driven")
plt.ylabel("Selling Price (Lakhs)")
plt.show()
```
